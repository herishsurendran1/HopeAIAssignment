```python
def OddEven():
    num = int(input("Enter a number : "))
    if(num % 2 == 0):
        print(num, " is an Even number")
    else:
        print(num, " is an Odd number")
```

```python
OddEven()
```

```python
def Eligible():
    gender = input("Your Gender:")
    age = int(input("Your Age:"))

    if(gender == 'Male' and age >=25):
        eligible = True
    elif(gender == 'Female' and age >=21):
        eligible = True
    else:
        eligible = False
    return eligible
```

```python
isEligible = Eligible()

if(isEligible == True):
    print("ELIGIBLE")
else:
    print("NOT ELIGIBLE")
```
